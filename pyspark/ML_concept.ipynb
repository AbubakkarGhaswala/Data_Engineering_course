```python
# import pyspark
import pyspark
```

```python
# import sparksession and using sql
from pyspark.sql import SparkSession
```

```python
# A must thing create an appname
spark = SparkSession.builder.appName("ML").getOrCreate()
```

```python
# import a csv file and inferschema true so we will get the original datatype
data = spark.read.csv('tips.csv', header=True, inferSchema=True)
```

```python
# this will display the data
data.show()
```

```python
# This printschema function will display the datatypes of data in the csv
data.printSchema()
```

```python
# Now we want to train a model so for that we need all data in numeric so we will convert the string data into numeric so we havae a
# function name StringIndexer so for that we have to use pyspark.ml and import StringIndexer
from pyspark.ml.feature import StringIndexer
```

```python
# Here this is a syntax of convert datatype for a single col
indexer=StringIndexer(inputCol="sex",outputCol="sex_indexed")
df_r=indexer.fit(data).transform(data)
df_r.show()
```

```python
# Here this is a syntax of convert datatype for a multiple col
indexer=StringIndexer(inputCols=["sex","smoker","day","time"],outputCols=["sex_indexed","smoker_indexed","day_indexed",
                                                                  "time_index"])
df=indexer.fit(data).transform(data)
df.show()
```

```python
# okay now what we have do is we have used vectore assembler so what he is do is we give multiple columns like tips,size,sex_indexed
# okay so it will take all data of the col and store in one col seprated by commas
from pyspark.ml.feature import VectorAssembler
featureassembler = VectorAssembler(inputCols=['tip','size','sex_indexed','smoker_indexed','day_indexed',
                          'time_index'],outputCol="Independent Features")
output=featureassembler.transform(df)
```

```python
output.select("Independent Features").show()
```

```python
output.show()
```

```python
finalized_data = output.select("Independent Features","total_bill")
```

```python
finalized_data.show()
```

```python

from pyspark.ml.regression import LinearRegression
##train test split
train_data,test_data=finalized_data.randomSplit([0.75,0.25])
regressor=LinearRegression(featuresCol='Independent Features', labelCol='total_bill')
regressor=regressor.fit(train_data)
```

```python
### Predictions
pred_results=regressor.evaluate(test_data)
```

```python
### PErformance Metrics
pred_results.r2,pred_results.meanAbsoluteError,pred_results.meanSquaredError
```
