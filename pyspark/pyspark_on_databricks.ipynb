```python
data = [(1,"Jhon","IT",50000),(2,"Ravi","HR",60000),(3,"Priya","IT",55000),(4,"Ravi","IT",65000),(5,"Priya","HR",70000)]

columns = ["id","name","dept","salary"]

df = spark.createDataFrame(data,columns)

df.show()
```

```python
df.printSchema()
```

```python
df.select("name","salary").show()
```

```python
df.filter(df.salary > 60000).show()
```

```python

df.filter((df.dept == "IT") & (df.salary > 55000)).show()
```

```python
df.filter((df.dept == "IT") | (df.salary > 55000)).show()
```

```python
df.groupBy("dept").sum("salary").show()
```

```python
df.orderBy(df.salary.desc()).show()
```

```python
result = df.collect()
result
```
