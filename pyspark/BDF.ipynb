```python
# this command will import pyspark - pyspark is use for big data and pandas use for small data
import pyspark
```

```python
# for use pyspark we have to create a spark session so below is the command
from pyspark.sql import SparkSession
```

```python
# for do further process we have to build a app so below is the command
spark=SparkSession.builder.appName("Demo").getOrCreate()
```

```python
# this command will goinf to import csv
df_pyspark = spark.read.csv('test1.csv',header=True,inferSchema=True)
```

```python
# this command is use to display the data in csv
df_pyspark.show()
```

```python
# this command will show the schema of the data which is inside the csv
df_pyspark.printSchema()
```

```python
# Selecting single column

df_pyspark.select("Name").show()
```

```python
# Selecting multiple column

df_pyspark.select(['Name','age']).show()
```

```python
# Adding column in Dataframe

df_pyspark=df_pyspark.withColumn('Experience After 2 Years',df_pyspark['Experience']+2)
```

```python
df_pyspark.show()
```

```python
# Dropping column

df_pyspark=df_pyspark.drop('Experience After 2 Years')
```

```python
df_pyspark.show()
```

```python
# For Rename The column
df_pyspark = df_pyspark.withColumnRenamed('Name','New_Name').show()
```
