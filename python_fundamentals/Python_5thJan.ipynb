## User defined datatypes
- Lists
- Tuples
- Sets
- Dictionaries

1. All of them store multiple values, heterogenous(the values stored can be of any datatype)
2. Tuples are like strings. They are immutable.
3. Lists are datatypes which are mutable, add/update/remove elements as we wish
4. Lists and tuples are ordered. The way you create them, that order is going to be there. So lists and tuples are ordered and hence indexible
5. Lists and tuples can have duplicates.
6. Sets are unordered, so no concept of index. Do not allow duplicates

Students in a class: list
Student who made enquiry on the website: tuple, geographical coordinates

All the 4 datatypes - lists, sets, tuples and dictionaries are heterogenous
- Lists: list() or []
    - Mutable, ordered(indexed)
    - append, insert, extend, remove, pop, Slicing/indexing, clear, count, index, in
- Tuples: tuple() or () or 2,3,4
    - Immutable, ordered(indexed)
    - count, index, in
- Set: set(), {2,3}
    - Mutable, unordered(no index), no duplicates
    - add, pop, remove, union, intersection, difference
- Dictionary: dict(), {}, {key:value, key:value}
    - Mutable, ordered(from 3.7 - no index)
    - pop, keys, values, items
### Nested Datatypes
- Can a list, set, tuple, dictionary exist inside a list?  YES
- Can a list, set, tuple, dictionary exist inside a tuple? NO
- Can a list, set, tuple, dictionary exist inside a set? YES
- Can a list, set, tuple, dictionary exist inside a dictionary?  YES

```python
# Branching - Conditional statements
# I want to award a grade based on your score. 
score = int(input('Enter your score'))
# If score is <35, your grade should be F. Else, it should be P
if score<35:
    print('Fail')
elif score<50:
    print('Pass')
elif score<70:
    print('Good')
else:
    print('Excellent')

# Few things to remember are:
# 1. Else is not compulsory. A conditional statement can exist with only if or just if and elif. 
# 2. Indentation - 1 tab press or 4 spaces is what python calls a standard indent. The level of indent tells which if-else branch you belong to.
# 3. Without if you cannot have elif or else. 
```

```python
# You get the score of students in 3 subjects. 
# The scores must be valid, meaning, it cannot be <0 or >100
# If the student scores <35 in any of the subject, he is considered fail
# If the student is pass, only then you check the average score.
# If average is more than 50, and if the student has >90 in any subject, he is considered exceptional
# Else, he is considered pass.

score1 = int(input('Enter your score'))
score2 = int(input('Enter your score'))
score3 = int(input('Enter your score'))
if score1<0 or score2<0 or score3<0 or score1>100 or score2>100 or score3>100:
    print('Error')
elif score1<35 or score2<35 or score3<35:
    print('Fail')
elif (score1+score2+score3)/3>50 and (score1>90 or score2>90 or score3>90):
    print('Exceptional')
elif (score1+score2+score3)/3>50:
    print('Good')
else:
    print('Pass')
```

```python
score1 = int(input('Enter your score'))
score2 = int(input('Enter your score'))
score3 = int(input('Enter your score'))
if score1<0 or score2<0 or score3<0 or score1>100 or score2>100 or score3>100:
    print('Error')
else:
    if score1<35 or score2<35 or score3<35:
        print('Fail')
    else:
        avg = (score1+score2+score3)/3
        if avg>50:
            if score1>90 or score2>90 or score3>90:
                print('Exceptional')
            else:
                print('Good')
        else:
            print('Pass')
```
