```python
#🟡 WHILE LOOP PRACTICE

#Print numbers from 1 to 5 using while.

numm = 0

while numm != 5:
    print(numm)
    numm = numm + 1

print(numm)
```

```python
# 🟠 IF + WHILE COMBINATION

# Print numbers from numbers list that are greater than 10.

numbers = [3, 7, 10, 15, 20, 25]

i = 0

while i < len(numbers):
    if numbers[i] > 10:
        print(numbers[i])
        i = i + 1
    else :
        i = i + 1
        
```

```python
# Print numbers from numbers list that are divisible by 5.

numbers = [3, 7, 10, 15, 20, 25]

i = 0

while i < len(numbers):
    if numbers[i] % 5 == 0:
        print(numbers[i])
        i = i + 1
    else:
        i = i + 1
```

```python
def sum_only_numbers():
    mixed_data = [10, "error", 20, "null", 30]
    total = 0
    for i in mixed_data:
        if type(i) == int:
            total = total + i
    return total

print(sum_only_numbers())
```

```python
 # Pass the list as a parameter 'data'
def sum_only_numbers(data):
    total = 0
    for i in data:
        if type(i) == int:
            total = total + i
    return total

# Now you can use it for ANY list!
mixed_list_1 = [10, "error", 20, "null", 30]
print(sum_only_numbers(mixed_list_1)) 

mixed_list_2 = [5, "junk", 5, 100, "skip"]
print(sum_only_numbers(mixed_list_2))
```

```python
def find_biggest(numbers):
    biggest = numbers[0]
    for i in numbers:
        if i > biggest:
            biggest = i
    return biggest

numbers = [3, 41, 12, 9, 74, 15]
print(find_biggest(numbers))
```

```python
def remove_dups(data):
    unique_list = []

    for i in data:
        if i not in unique_list:
            unique_list.append(i)
    return unique_list

data = [1, 2, 2, 3, 4, 4, 4, 5]
print(remove_dups(data))
```

```python
def count_occurrences(data,search):
    counter = 0
    
    for i in data:
        if i == search:
            counter = counter + 1
    return counter

data = ["Apple", "Banana", "Apple", "Orange", "Apple"]
search = "Apple"

print(count_occurrences(data,search))
```

```python
def filter_high_prices(prices,threshold):
    filtered_list = []


    for i in prices:
        if i > threshold:
            filtered_list.append(i)

    return sorted(filtered_list)

prices = [10, 55, 20, 100, 5, 80]
threshold = 50

print(filter_high_prices(prices,threshold))
```

```python
staff = [
    {"name": "Ali", "dept": "Sales", "salary": 2000},
    {"name": "John", "dept": "IT", "salary": 3000},
    {"name": "Abubakkar", "dept": "Sales", "salary": 5000}
]

def process_bonuses(staff):
    results_list = []

    for i in staff:
        # 1. Logic to determine the bonus amount
        if i["dept"] == "Sales":
            bonus = i["salary"] * 0.10
        else:
            bonus = 0
            
        # 2. These MUST be inside the loop to happen for every person!
        ans = f"Name: {i['name']}, Bonus: {bonus}"
        results_list.append(ans)
        
    return results_list

print(process_bonuses(staff))
```
