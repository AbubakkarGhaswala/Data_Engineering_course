```python
## Aggregation on a Database

data.groupby('Sex').describe()
```

```python
data.groupby('Sex').Age.mean()
```

```python
dff = data.groupby('Pclass').agg({
    "Age" : ['min','max','mean'],
    "Fare" : ['min','max','mean'],
    "Survived" : ['sum','count']
})
dff[('non_survived')] = dff[('Survived','count')] - dff[('Survived','sum')]
dff
```
