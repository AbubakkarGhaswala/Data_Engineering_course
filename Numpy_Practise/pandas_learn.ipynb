### Pandas is a library that helps you handle tabular data
- Data in Pandas is stored as dataframes, a 2-d structure similar to the tables in excel
- Pandas offers transformations, data cleaning, preprocessing options, and hence pandas library becomes an eminent part of life of DE and DS
- Pandas offers two data abstraction models - Series and DataFrame

```python
#Series in pandas is a 1-d data. Dataframe is a 2-d data. Other than that we do not have any dimensional data in pandas
```

```python
import pandas as pd
#A series can be constructed from a list, tuple, a np.1-d array, a dictionary. But a series cannot be constructed from a set.
a = pd.Series([1,2,3,4])
a
```

```python
ser1 = pd.Series([3,4,5,6], index = ['row1','row2','row3','row4'], name = "Series1")
ser1
```

```python
df = pd.DataFrame([[1,2,3,4,5],[5,6,7]])
df
```

```python
df2=pd.DataFrame(ser1)
df2
```

```python
df = pd.DataFrame([[1,2,3],[6,7,8],[9,4,3,2,6,8]], index=['row1','row2','row3'], columns = ['col1','col2','col3','col4','col5','col6'])
df
```

```python
df.rename(columns = {'col1':'c1'}, inplace=True)   #inplace=True ensures that a permanent change is done in the variablef
```
