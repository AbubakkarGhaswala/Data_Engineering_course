```python
## First We have import pandas library and use it as pd 
## then i have use read_csv function of pandas to import csv file and print it
import pandas as pd

data = pd.read_csv("Student_Data_Practice.csv")
data
```

```python
## Before do anything i have drop the duplicates from data for better work
data.drop_duplicates()
```

```python
## Then i took a small tour of our data to know datatypes and little details about our dataset
data.info()
```

```python
## Then i have use use isnull with sum function to find out how many null values we have in our dataset
data.isnull().sum()
```

```python
## Then i got 25 null cites entry in dataset so i have fill it with unknown 
data['City'] = data['City'].fillna("Unknown")
data.to_csv("Student_Data_Practice.csv",index=False)
```

```python
## Then i have found 25 around null values in maths marks col so i found out the mean of math marks and fill null values with mean number
mean_of_Math = data['Math'].mean()
data['Math'] = data['Math'].fillna(mean_of_Math)

data.to_csv("Student_Data_Practice.csv",index=False)
```

```python
## Then i have found 25 around null values in Science marks col so i found out the mean of Science marks and fill null values with mean number

mean_of_Science = data['Science'].mean()
data['Science'] = data['Science'].fillna(mean_of_Science)

data.to_csv("Student_Data_Practice.csv",index=False)
```

```python
data.rename(columns={'Attendance_%': 'Attendance'}, inplace=True)
data.to_csv("Student_Data_Practice.csv",index=False)
```
